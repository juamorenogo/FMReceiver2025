```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Funciones generales

def db_to_linear(db):
    return 10 ** (db / 10)

def linear_to_db(linear):  
    return 10 * np.log10(linear)

def paralelo(R1, R2):
    return (R1*R2)/(R1 + R2)
```

```python
# Variables polarizacion DC
Av = -7.5  # Ganancia lineal deseada
Ic = 9e-3 # A
vcc = 8 # V
vce = 3 # V
vbe = 0.7 # V
beta = 100
R_L = 50  # Ohm
print("---------**---------")
print("Punto de operación:")
print(f"Ic = {Ic*1e3} mA")
print(f"Vce = {vce} V")


# Variables adicionales calculadas
alpha = beta/(beta + 1)
Ie = Ic/alpha
Ib = Ic/beta
gm = Ic / 25e-3  # Transconductancia

print("---------**---------")
# Calculo de resistencia equivalente
R_eq = -Av/gm
print("R_eq:", R_eq)


Rc = R_eq*R_L/(R_L - R_eq)

# Definir símbolos
r1, r2, va = sp.symbols('r1 r2 va')

# Sistema de ecuaciones
eq1 = sp.Eq(vcc - va, r1 * Ie)
eq2 = sp.Eq(va - vce, Rc * Ic)
eq3 = sp.Eq(va - vbe, r2 * Ib)

print("---------**---------")
# Resolver para r1 y r2
sol_r1_r2 = sp.solve((eq1, eq2, eq3), (r1, r2, va))
print("Solución completa:", sol_r1_r2)

# O si solo quieres r1 y r2:
r1_val = sol_r1_r2[r1]
r2_val = sol_r1_r2[r2]
va_val = sol_r1_r2[va]

print("---------**---------")
print(f"Rp1 = {r2_val:.2f} Ω") 
print(f"Rp2 = {r1_val:.2f} Ω")
print(f"Rc = {Rc:.2f} Ω")
print(f"va aprox = {va_val:.2f} V")
```
